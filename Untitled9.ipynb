```python
import pandas as pd

file_path = '/content/drive/MyDrive/data/car.xlsx'

df = pd.read_excel(file_path)

df.head()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace(r'[^\w]', '', regex=True)
)
```

```python
df.columns
```

```python
df['name'] = df['name'].fillna('abc...')
```

```python
df['address'] = df['address'].fillna('not_applicable')
```

```python
df['assettype'] = df['assettype'].fillna('unknown')
```

```python
df.duplicated().sum()
```

```python
df.info()
```

```python
p = ['name', 'address', 'city', 'submodel', 'model', 'assettype', 'misstatus', 'promotiondesc', 'product']
for col in p :
  df[col] = df[col].astype(str).str.strip().str.lower()
```
