```python
import pandas as pd

file_path = '/content/drive/MyDrive/data/hinduism_860288.xlsx'

df = pd.read_excel(file_path)

df.head()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace(r'[^\w]', '', regex=True)
)
```

```python
df.columns
```

```python
df['f_name'] = df['f_name'].fillna('abc..')
```

```python
df['email'] = df['email'].fillna('example@gmail.com')
```

```python
df['mobile'] = df['mobile'].fillna('na')
```

```python
df['acno'] = df['acno'].fillna('unknown')
df['ifscode'] = df['ifscode'].fillna('unknown')
df['banknm'] = df['banknm'].fillna('unknown')
```

```python
df.isnull().sum()
```

```python
df['cadd'] = df['cadd'].fillna('unknown')
```

```python
df['ccity'] = df['ccity'].fillna('unknown')
```

```python
df['gdate'] = df['gdate'].fillna('mm dd yyyy hh:mm AM|PM')
```

```python
df.isnull().sum()
```

```python
df['m_name'] = df['m_name'].fillna('abc...')
```

```python
df.isnull().sum()
```

```python
df.head(3)
```

```python
df.columns
```

```python
p = ['c_name', 'f_name', 'm_name', 'gender', 'caste', 'religion', 'ccity', 'centre_nam', 'dt_name', 'cdistt']
for col in p :
  df[col] = df[col].astype(str).str.strip().str.lower()
```
